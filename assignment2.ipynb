## Part III: Functions

### 1. Write a function to convert temperature from kelvin to celsius and celsius to kelvin

````python
def convert_k_c(temperature, input_unit="k"):

    unit = input_unit.lower()

    if unit in ("k"):
        if temperature < 0:
            raise ValueError("Kelvin temperature cannot be negative.")
        return temperature - 273.15
    if unit in ("c"):
        kelvin = temperature + 273.15
        if kelvin < 0:
            raise ValueError("The temperature cannot be lower than absolute zero.")
        return kelvin

    raise ValueError("input_unit must be 'k' or 'c'.")
````

### 2. Write a function to convert temperature to fahrenheit

Include an optional keyword argument to specify whether the input is in  celcius or kelvin.
Call your previously defined functions if necessary.

````python
def to_fahrenheit(temperature, input_unit="c"):

    unit = input_unit.lower()

    if unit in ("k"):
        celsius = convert_k_c(temperature, input_unit="k")
    elif unit in ("c"):
        celsius = temperature
    else:
        raise ValueError("input_unit must be 'c' or 'k'.")

    return celsius * 9 / 5 + 32
````

### 3) Check that the outputs are sensible

by trying a few examples

````python
print(to_fahrenheit(33))
print(to_fahrenheit(310, input_unit="k"))
print(to_fahrenheit(0))
print(to_fahrenheit(100))
print()
print(convert_k_c(273.15))
print(convert_k_c(300))
print(convert_k_c(0, input_unit="c"))
print(convert_k_c(100, input_unit="c"))
````

### 4) Now write a function that converts _from_ farenheit

and uses a keyword argument to specify whether you want the output in celcius or kelvin

````python
def from_fahrenheit(temperature, output_unit="c"):
    celsius = (temperature - 32) * 5 / 9
    unit = output_unit.lower()

    if unit in ("c"):
        return celsius
    if unit in ("k"):
        return convert_k_c(celsius, input_unit="c")

    raise ValueError("output_unit must be 'c' or 'k'.")
````

### 5) Write a function that takes two arguments (feet and inches) and returns height in meters

Verify it gives sensible answers

````python
def feet_inches_to_meters(feet, inches):
    
    if feet < 0 or inches < 0:
        raise ValueError("Length cannot be negative.")

    total_inches = feet * 12 + inches
    return total_inches * 0.0254
````

### 6. Write a function takes one argument (height in meters) and returns two arguments (feet and inches)

(Consult the [tutorial on numbers](https://docs.python.org/3/tutorial/introduction.html#numbers) if you are stuck on how to implement this.)

````python
def meters_to_feet_inches(meters):

    if meters < 0:
        raise ValueError("Length cannot be negative.")

    total_inches = meters / 0.0254
    feet = int(total_inches // 12)
    inches = total_inches - feet * 12
    return feet, inches
````

### 7) Verify that the "round trip" conversion from and back to meters is consistent
